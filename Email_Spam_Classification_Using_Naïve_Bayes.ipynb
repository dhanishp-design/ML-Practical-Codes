```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = {
    'message': [
        'Win a free prize now',
        'Claim your free cash reward',
        'Congratulations you won a lottery',
        'Get cheap loans now',
        'Free entry in a contest',
        'You have won free money',
        'Meeting is scheduled for tomorrow',
        'Please submit your assignment',
        'Let us study machine learning',
        'Can we meet today',
        'Your class starts at ten',
        'Please send me the notes',
        'Project submission is tomorrow',
        'Are you coming to college',
        'The lecture is very helpful',
        'Win cash and claim your prize',
        'Free gift waiting for you',
        'Call now to claim your reward',
        'Please check your email',
        'We have a meeting today'
    ],

    'label': [
        'spam', 'spam', 'spam', 'spam', 'spam',
        'spam', 'ham', 'ham', 'ham', 'ham',
        'ham', 'ham', 'ham', 'ham', 'ham',
        'spam', 'spam', 'spam', 'ham', 'ham'
    ]
}

df = pd.DataFrame(data)

df.head()
```

```python
X = df['message']
y = df['label']

print(X.head())
print(y.head())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
vectorizer = CountVectorizer()

X_train_vector = vectorizer.fit_transform(X_train)
X_test_vector = vectorizer.transform(X_test)
```

```python
model = MultinomialNB()

model.fit(X_train_vector, y_train)

print("Model training completed!")
```

```python
y_pred = model.predict(X_test_vector)

print("Actual labels:")
print(list(y_test))

print("Predicted labels:")
print(list(y_pred))
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy * 100, "%")

print(classification_report(
    y_test,
    y_pred,
    zero_division=0
))
```

```python
email = ["Congratulations! You have won free cash"]

email_vector = vectorizer.transform(email)

prediction = model.predict(email_vector)

if prediction[0] == 'spam':
    print("This email is SPAM")
else:
    print("This email is NOT SPAM")
```
